```python
y=df["Churn"]
x=df[["Age","Gender","Tenure","MonthlyCharges"]]
```

```python
y=df["Churn"].apply(lambda x:1 if x=="Yes" else 0)
```

```python
y.head()
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
x
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
```

```python
import joblib
joblib.dump(scaler,"scaler.pkl")
```

```python
X_test=scaler.fit_transform(X_test)
```

```python
from sklearn.metrics import accuracy_score
def modelper(predictions):
    print("Accuracy score of the model is {} ".format(accuracy_score(y_test,predictions)))
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
log_model=LogisticRegression()
```

```python
log_model.fit(X_train,y_train)
```

```python
y_pred=log_model.predict(X_test)
```

```python
modelper(y_pred)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
param_grid={
    "n_neighbors":[3,5,7,9],
    "weights": ["uniform","distance"]
}
```

```python
girdkn=GridSearchCV(KNeighborsClassifier(),param_grid,cv=5)
```

```python
girdkn.fit(X_train,y_train)
```

```python
girdkn.best_params_
```

```python
y_pred=girdkn.predict(X_test)
```

```python
modelper(y_pred)
```

```python
from sklearn.svm import SVC
```

```python
svm= SVC()
```

```python
param_grid={
"C": [0.01,0.1,0.5,1],
"kernel":["linear","rbf","poly"]
}
```

```python
gridsvc=GridSearchCV(svm,param_grid,cv=5)
```

```python
gridsvc.fit(X_train,y_train)
```

```python
gridsvc.best_params_
```

```python
y_pred=gridsvc.predict(X_test)
```

```python
modelper(y_pred)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
param_grid={
    "criterion":["gini","entropy"],
    "splitter":["best","random"],
    "max_depth":[None,10,20,30],
    "min_samples_split":[2,5,10],
    "min_samples_leaf":[1,2,4]

}
```

```python
grid_tree=GridSearchCV(DecisionTreeClassifier(),param_grid,cv=5)
```

```python
grid_tree.fit(X_train,y_train)
```

```python
grid_tree.best_params_
```

```python
y_pred=grid_tree.predict(X_test)
```

```python
y_pred
```

```python
modelper(y_pred)
```
